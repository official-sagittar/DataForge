```python
iqr_stats = (
    quite_df
    .groupby(["pos_phase_label", "game_wdl"])["pos_eval_white_pov"]
    .agg(
        q1=lambda s: s.quantile(0.25),
        q3=lambda s: s.quantile(0.75),
    )
    .reset_index()
)
```

```python
iqr_stats["iqr"] = iqr_stats["q3"] - iqr_stats["q1"]
iqr_stats["lower_bound"] = iqr_stats["q1"] - 1.5 * iqr_stats["iqr"]
iqr_stats["upper_bound"] = iqr_stats["q3"] + 1.5 * iqr_stats["iqr"]
```

```python
iqr_stats
```

```python
work_df = quite_df.merge(
    iqr_stats,
    on=["pos_phase_label", "game_wdl"],
    how="left",
)
```

```python
work_df
```

```python
work_df["is_outlier"] = ~work_df["pos_eval_white_pov"].between(
    work_df["lower_bound"],
    work_df["upper_bound"],
)
```

```python
clean_df = work_df[~work_df["is_outlier"]].copy()
```

```python
medians = (
    clean_df
    .groupby(["pos_phase_label", "game_wdl"])["pos_eval_white_pov"]
    .median()
    .reset_index()
    .rename(columns={"pos_eval_white_pov": "phase_result_median_eval"})
)

display(medians)
```

```python
clean_df = clean_df.merge(
    medians,
    on=["pos_phase_label", "game_wdl"],
    how="left",
)
```

```python
DRAW_BAND = 0.05
```

```python
clean_df["is_signal"] = (
    (
        (clean_df["game_wdl"] == 1.0)
        & (clean_df["pos_eval_white_pov"] > clean_df["phase_result_median_eval"])
    )
    |
    (
        (clean_df["game_wdl"] == 0.0)
        & (clean_df["pos_eval_white_pov"] < clean_df["phase_result_median_eval"])
    )
    |
    (
        (clean_df["game_wdl"] == 0.5)
        & clean_df["pos_eval_white_pov"].between(
            clean_df["phase_result_median_eval"] - DRAW_BAND,
            clean_df["phase_result_median_eval"] + DRAW_BAND,
        )
    )
)
```

```python
training_df = clean_df[clean_df["is_signal"]].copy()
```

```python
training_df
```

```python
training_summary = (
    training_df
    .groupby(["pos_phase_label", "game_wdl"])["pos_eval_white_pov"]
    .agg(["count", "mean", "median", "std", "min", "max"])
    .reset_index()
)

display(training_summary)
```

```python
import pandas as pd

def sample_uniform_phase_wdl(
    df: pd.DataFrame,
    n: int,
    random_state: int = 42,
    replace: bool = False,
) -> pd.DataFrame:
    group_cols = ["pos_phase_label", "game_wdl"]

    # Number of joint groups, e.g. EG/MG x 0/0.5/1 = 6
    groups = df.groupby(group_cols, dropna=False)
    n_groups = groups.ngroups

    base_n = n // n_groups
    remainder = n % n_groups

    sampled_parts = []

    # Stable ordering so remainder allocation is reproducible
    group_items = list(groups)
    group_items = sorted(group_items, key=lambda x: str(x[0]))

    for i, (group_key, group_df) in enumerate(group_items):
        target_n = base_n + (1 if i < remainder else 0)

        if len(group_df) < target_n and not replace:
            raise ValueError(
                f"Group {group_key} has only {len(group_df)} rows, "
                f"but target sample size is {target_n}. "
                f"Use replace=True or reduce n."
            )

        sampled_parts.append(
            group_df.sample(
                n=target_n,
                replace=replace,
                random_state=random_state + i,
            )
        )

    sampled_df = (
        pd.concat(sampled_parts, ignore_index=True)
        .sample(frac=1, random_state=random_state)
        .reset_index(drop=True)
    )

    return sampled_df
```

```python
N = 50_000

sample_df = sample_uniform_phase_wdl(
    training_df,
    n=N,
    random_state=42,
    replace=False,
)
```
